Import libraries

```python
import pandas as pd
import numpy as np
```

Load dataset

```python
df = pd.read_csv("../datasets/ecommerce_product_return_prediction.csv")

print("Dataset shape:", df.shape)
df.head()
```

Fix data types

```python
df["order_date"] = pd.to_datetime(df["order_date"], errors="coerce")
df["return_date"] = pd.to_datetime(
    df["return_date"].replace("Not Returned", pd.NA),
    errors="coerce"
)

numeric_columns = [
    "customer_age", "product_price_usd", "discount_percent",
    "product_rating", "past_purchase_count", "past_return_rate_pct",
    "session_length_minutes", "num_product_views", "delivery_time_days",
    "delivery_delay_days", "total_orders_lifetime",
    "total_returns_lifetime", "return_rate_pct", "returned",
    "days_to_return", "refund_amount_usd"
]

df[numeric_columns] = df[numeric_columns].apply(
    pd.to_numeric, errors="coerce"
)

print(df.dtypes)
```

Check invalid values

```python
print("Invalid prices:", (df["product_price_usd"] <= 0).sum())
print("Invalid ratings:", (~df["product_rating"].between(1, 5)).sum())
print("Invalid discounts:", (~df["discount_percent"].between(0, 100)).sum())
print("Invalid target values:", (~df["returned"].isin([0, 1])).sum())
```

Remove columns not useful for pre-return prediction

```python
remove_columns = [
    "order_id",
    "customer_id",
    "return_date",
    "days_to_return",
    "return_reason",
    "refund_amount_usd"
]

df = df.drop(columns=remove_columns, errors="ignore")

print("Remaining columns:")
print(df.columns.tolist())
print("Dataset shape:", df.shape)
```

Separate features and target

```python
X = df.drop(columns=["returned"])
y = df["returned"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("\nTarget distribution:")
print(y.value_counts())
```

Check the processed dataset

```python
print("Dataset shape:", df.shape)
print("Duplicate rows:", df.duplicated().sum())
print("Missing values:", df.isnull().sum().sum())
print("Target values:", sorted(df["returned"].unique()))

df.head()
```

Save the processed dataset

```python
df.to_csv(
    "../datasets/ecommerce_return_prediction_preprocessed_clean.csv",
    index=False
)

print("Preprocessed dataset saved successfully.")
```
