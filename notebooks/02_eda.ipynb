Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

Load Dataset

```python
df = pd.read_csv("../datasets/ecommerce_product_return_prediction.csv")
df.head()
```

Return and non-return distribution

```python
print(df["returned"].value_counts())
print("\nReturn percentage:")
print((df["returned"].value_counts(normalize=True) * 100).round(2))
```

Target distribution chart

```python
sns.countplot(data=df, x="returned")
plt.title("Returned vs Non-Returned Orders")
plt.xlabel("Returned (0 = No, 1 = Yes)")
plt.ylabel("Number of Orders")
plt.show()
```

Product return analysis ||
Return rate by product category

```python
category_returns = df.groupby("product_category")["returned"].agg(
    ["mean", "count"]
)
category_returns["return_rate_pct"] = category_returns["mean"] * 100
category_returns = category_returns.sort_values("return_rate_pct")

category_returns["return_rate_pct"].plot(kind="bar", figsize=(9, 5))
plt.title("Return Rate by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Return Rate (%)")
plt.xticks(rotation=35)
plt.tight_layout()
plt.show()

category_returns
```

Product price and return status

```python
sns.boxplot(data=df, x="returned", y="product_price_usd")
plt.title("Product Price by Return Status")
plt.xlabel("Returned (0 = No, 1 = Yes)")
plt.ylabel("Product Price (USD)")
plt.show()
```

Product ratings and returns

```python
rating_returns = df.groupby("product_rating")["returned"].mean() * 100

rating_returns.plot(kind="bar", figsize=(8, 4))
plt.title("Return Rate by Product Rating")
plt.xlabel("Product Rating")
plt.ylabel("Return Rate (%)")
plt.show()
```

Discount percentage and returns

```python
df["discount_group"] = pd.cut(
    df["discount_percent"],
    bins=[-1, 10, 25, 40, 70],
    labels=["0–10%", "10–25%", "25–40%", "40%+"]
)

discount_returns = df.groupby(
    "discount_group",
    observed=False
)["returned"].mean() * 100

discount_returns.plot(kind="bar", figsize=(8, 4))
plt.title("Return Rate by Discount Percentage")
plt.xlabel("Discount")
plt.ylabel("Return Rate (%)")
plt.show()
```

Delivery time and returns

```python
sns.boxplot(data=df, x="returned", y="delivery_time_days")
plt.title("Delivery Time by Return Status")
plt.xlabel("Returned (0 = No, 1 = Yes)")
plt.ylabel("Delivery Time (Days)")
plt.show()
```

Return rate by payment method

```python
payment_returns = df.groupby("payment_method")["returned"].mean() * 100

payment_returns.plot(kind="bar", figsize=(9, 4))
plt.title("Return Rate by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Return Rate (%)")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

Return reasons

```python
returned_orders = df[df["returned"] == 1]

returned_orders["return_reason"].value_counts().plot(
    kind="bar", figsize=(9, 4)
)
plt.title("Reasons for Product Returns")
plt.xlabel("Return Reason")
plt.ylabel("Number of Orders")
plt.xticks(rotation=35)
plt.tight_layout()
plt.show()
```

Monthly return trends

```python
df["order_date"] = pd.to_datetime(df["order_date"])

monthly_returns = df.set_index("order_date").resample("MS")["returned"].mean() * 100

monthly_returns.plot(figsize=(10, 4), marker="o")
plt.title("Monthly Return Rate")
plt.xlabel("Month")
plt.ylabel("Return Rate (%)")
plt.show()
```
