Import libraries

```python
import pandas as pd
import numpy as np
```

Load preprocessed dataset

```python
df = pd.read_csv("../datasets/ecommerce_return_prediction_preprocessed_clean.csv")

print("Dataset shape:", df.shape)
df.head()
```

Create customer purchase groups

```python
df["purchase_frequency_group"] = pd.cut(
    df["past_purchase_count"],
    bins=[-1, 5, 15, 30, np.inf],
    labels=["Low", "Medium", "High", "Very High"]
)

df[["past_purchase_count", "purchase_frequency_group"]].head()
```

Create product price groups

```python
df["price_category"] = pd.cut(
    df["product_price_usd"],
    bins=[0, 25, 100, 300, np.inf],
    labels=["Budget", "Mid Range", "Premium", "Luxury"]
)

df[["product_price_usd", "price_category"]].head()
```

Create order month and day features

```python
df["order_date"] = pd.to_datetime(df["order_date"])

df["order_month"] = df["order_date"].dt.month
df["order_day_of_week"] = df["order_date"].dt.dayofweek
df["is_weekend"] = (df["order_day_of_week"] >= 5).astype(int)

df[["order_date", "order_month", "order_day_of_week", "is_weekend"]].head()
```

Create customer return history groups

```python
df["customer_return_history"] = pd.cut(
    df["past_return_rate_pct"],
    bins=[-1, 5, 15, 30, 100],
    labels=["Low", "Moderate", "High", "Very High"]
)

df[["past_return_rate_pct", "customer_return_history"]].head()
```

Analyze engineered features against returns

```python
print("Return rate by price category:")
print(
    (df.groupby("price_category", observed=False)["returned"].mean() * 100)
    .round(2)
)

print("\nReturn rate by purchase frequency:")
print(
    (df.groupby("purchase_frequency_group", observed=False)["returned"].mean() * 100)
    .round(2)
)
```

```python
df.to_csv(
    "../datasets/ecommerce_return_prediction_feature_engineered.csv",
    index=False
)

print("Dataset saved successfully:", df.shape)
```
